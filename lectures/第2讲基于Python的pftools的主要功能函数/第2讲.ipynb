```python
from parflow.tools.hydrology import calculate_subsurface_storage, calculate_surface_storage
from parflow.tools.hydrology import calculate_water_table_depth, calculate_overland_flow_grid
```

```python
specific_storage = read_pfb(get_absolute_path('../../examples/parflow-clm/outputs/PF_CLM_Example/PF_CLM_Example.out.specific_storage.pfb'))
porosity = read_pfb(get_absolute_path('../../examples/parflow-clm/outputs/PF_CLM_Example/PF_CLM_Example.out.porosity.pfb'))
mask10 = read_pfb(get_absolute_path('../../examples/parflow-clm/outputs/PF_CLM_Example/PF_CLM_Example.out.mask.pfb'))
# PME = read_pfb(get_absolute_path('./inputs/PME_flux.pfb'))
slopex = read_pfb(get_absolute_path('../../examples/parflow-clm/inputs/slope_x.pfb'))
slopey = read_pfb(get_absolute_path('../../examples/parflow-clm/inputs/slope_y.pfb'))
manning = read_pfb(get_absolute_path('../../examples/parflow-clm/inputs/manning.pfb'))

dx = 961.72
dy = 961.72
dz = np.array([300,100,50,25,10,5,1,0.6,0.3,0.1])
```

```python
saturation = read_pfb(get_absolute_path('../../examples/parflow-clm/outputs/PF_CLM_Example/PF_CLM_Example.out.satur.00062.pfb'))
pressure = read_pfb(get_absolute_path('../../examples/parflow-clm/outputs/PF_CLM_Example/PF_CLM_Example.out.press.00062.pfb'))
```

```python
streamflow = calculate_overland_flow_grid(pressure, slopex, slopey, manning, dx, dy, mask = mask10)[::-1]
```

```python
import matplotlib.pyplot as plt
import matplotlib
# %matplotlib widget
```

```python
fig = plt.figure(figsize=(16,10))

plt.subplot(121)
im = plt.imshow(streamflow,norm=matplotlib.colors.LogNorm(vmin=1e-2, vmax=1e5))
plt.colorbar(im, fraction=0.02, pad=0.04)
plt.title('streamflow ($\mathregular{m^3}$/h)')
```
