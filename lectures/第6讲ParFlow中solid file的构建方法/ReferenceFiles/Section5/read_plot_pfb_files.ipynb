```python
import numpy as np
import matplotlib.pyplot as plt
from parflow.tools.io import read_pfb
from parflow.tools.fs import get_absolute_path
```

```python
pfb_files = {
    'Basin bottom': 'Sabino_Basin_Bot.pfb',
    'Basin mask': 'Sabino_Basin_Mask.pfb',
    'Dummy mask': 'Sabino_Dummy_Mask.pfb',
    'Domain mask': 'Sabino_Mask.pfb',
    'DEM': 'Sabino_DEM.pfb',
    'MBGeo bottom': 'Sabino_MBGeo_Bot.pfb',
    'MBGeo mask': 'Sabino_MBGeo_Mask.pfb',
}

pfb_data = {
    name: read_pfb(get_absolute_path(f'./{filename}'))
    for name, filename in pfb_files.items()
}

{name: data.shape for name, data in pfb_data.items()}
```

```python
fig, axes = plt.subplots(2, 4, figsize=(20, 9), constrained_layout=True)

for ax, (name, data) in zip(axes.flat, pfb_data.items()):
    image = ax.imshow(data[0], origin='lower')
    ax.set_title(name)
    ax.set_xlabel('x index')
    ax.set_ylabel('y index')
    fig.colorbar(image, ax=ax, shrink=0.85)

for ax in axes.flat[len(pfb_data):]:
    ax.set_visible(False)

fig.suptitle('Sabino PFB input files', fontsize=16)
plt.show()
```
